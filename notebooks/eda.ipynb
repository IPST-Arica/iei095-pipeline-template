```python
import polars as pl
import duckdb
import plotly.express as px

# Conectar a la base DuckDB generada por dbt
con = duckdb.connect("../data/transformed/superstore.duckdb", read_only=True)

# Consultar la fact table
df = con.execute("SELECT * FROM analytics.fct_sales_summary").pl()

# Visualización: ventas por región y mes
fig = px.bar(
    df.to_pandas(),
    x="month", y="total_sales",
    color="region",
    title="Ventas totales por mes y región",
    labels={"total_sales": "Ventas (USD)", "month": "Mes"}
)
fig.show()

# Top 5 categorías por rentabilidad (profitability)
profit_by_cat = (
    df.group_by("category")
    .agg(pl.col("total_profit").sum())
    .sort("total_profit", descending=True)
)
print(profit_by_cat)
```
